```python
import numpy as np
#a
stock1=np.genfromtxt('stocks1.csv',encoding='UTF-8',names=True,delimiter=',',dtype=float)
print(stock1)

print('5 dong dau:\n',stock1[:5])
print('5 dong cuoi:\n',stock1[-5:])
print('kieu du lieu cua cac cot:\n',stock1.dtype.fields)

print('thong tin:\n',np.info(stock1))
```

```python
stock2=np.genfromtxt('stocks2.csv',encoding='UTF-8',names=True,delimiter=',')
print('5 dong dau:\n',stock2[:5])
print('5 dong cuoi:\n',stock2[-5:])
print('kieu du lieu cua cac cot:\n',stock2.dtype.fields)
print('thong tin:\n',np.info(stock2))
```

```python
companies=np.genfromtxt('companies.csv',encoding='UTF-8',names=True,delimiter=',')
print(companies)

print('kieu du lieu cua cac cot:\n',companies.dtype.fields)

print('thong tin:\n',np.info(companies))
```
